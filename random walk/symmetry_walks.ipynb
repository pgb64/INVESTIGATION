```python
# Importamos la librerías necesarias

import numpy as np
import os, csv
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def iteration(i: int, n: int) -> None:
    """
    Muestra en una sola línea la iteración actual sin imprimir nuevas líneas.
    
    Parámetros:
    - n (int): Número total de iteraciones.
    """
    print(f"\rIteración: {i + 1}/{n}", end = "", flush = True)
```

```python
def hitting_time(dims : int, size : int, n : int = 1000) -> tuple[float, float, int]:
    """
    Calcula el hitting time para un Random Walk en 'dims' dimensiones.

    Parámetros:
    - dims (int): Dimensión del espacio (1D, 2D, 3D...).
    - size (int): Límite del espacio (el walker empieza en size // 2).
    - n (int): Número de simulaciones (default = 10,000).

    Retorna:
    - (media_hitting_time, std_hitting_time, max_hitting_time)
    """
    start = size // 2  
    options = np.eye(dims, dtype = int)  
    options = np.vstack([options, -options])  

    hitting_times = np.zeros(n, dtype = int)

    for i in range(n):
        # iteration(i, n)
        pos = np.full(dims, start)
        steps = 0

        while np.all(pos > 0) and np.all(pos < size):
            pos += options[np.random.randint(0, 2 * dims)]
            steps += 1
        
        hitting_times[i] = steps 

    return dims, size, np.mean(hitting_times), np.std(hitting_times), np.max(hitting_times)
```

```python
def guardar_csv(data : tuple[int, int, float, float, float], filename : str) -> None:
    """
    Guarda los resultados en un archivo CSV
    
    Parámetros:
    - data (tuple): Datos a guardar en formato (dims, size, hitting, std, max).
    - filename (str): Nombre del archivo CSV.
    """

    filepath = f"../data/{filename}"  
    file_exists = os.path.exists(filepath)

    existing_data = set()
    if file_exists:
        with open(filepath, mode = 'r', newline = '') as file:
            reader = csv.reader(file)
            next(reader, None)  
            existing_data = {tuple(row) for row in reader} 

    new_data = (f"{data[0]}D", str(data[1]), str(data[2]), str(data[3]), str(data[4]))

    if new_data in existing_data:
        return

    with open(filepath, mode = 'a', newline = '') as file:
        writer = csv.writer(file)
        if not file_exists:
            writer.writerow(["Dimension", "Size", "Hitting Time", "STD", "Max"])

        writer.writerow(new_data)
```

```python
sizes = [2 ** s for s in range(2, 8)] 
dims = [1, 2, 3, 4, 5, 6, 8, 12, 16, 32, 64, 128]
```

```python
# Prueba con distintas semillas
for d in dims:
    print(f"\n{d}D\n---------------")
    for s in sizes: 
        results = []
        for seed in range(10):
            np.random.seed(seed)
            results.append(hitting_time(d, s))

        mean_hitting = np.mean([d[2] for d in results])
        mean_std = np.mean([d[3] for d in results])
        mean_max = np.mean([d[4] for d in results]) 
        
        print(f'Completed: size {s}')
        guardar_csv((d, s, mean_hitting, mean_std, mean_max), "../data/symmetry_walks.csv")
```
