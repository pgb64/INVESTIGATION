# DRUNK WALK

```python
# Importamos los módulos necesarios
import numpy as np
import matplotlib.pyplot as plt
import random, time, os, csv
```

```python
def iteraciones(i : int, n : int) -> None:
    '''
    Function that show the progress

    Parameters:
    -----------
    i : int 
        Iteration number
    n : int
        Total number of iterations

    Returns:
    --------
    None
    '''
    print(f"\r{i + 1}/{n}", end="", flush=True)
```

```python
def random_walk_ND(n: int, a: tuple, b: tuple) -> tuple[float, float]:
    '''
    Function that generates a symmetric random walk in N dimensions

    Parameters:
    -----------
    n : int
        Number of walks
    a : tuple
        Minimum boundary in each dimension (size N)
    b : tuple
        Maximum boundary in each dimension (size N)

    Returns:
    --------
    tuple[float, float]
        Mean number of steps and max steps in a single walk
    '''
    dimensions = len(a)  # Number of dimensions
    steps = []
    
    for _ in range(n):
        iteraciones(_, n)
        init = [(a[i] + b[i]) // 2 for i in range(dimensions)]  # Start in the middle
        moves = 0

        while all(a[i] < init[i] < b[i] for i in range(dimensions)):  
            move = random.choice([tuple(1 if i == j else 0 for i in range(dimensions)) for j in range(dimensions)] +
                                 [tuple(-1 if i == j else 0 for i in range(dimensions)) for j in range(dimensions)])
            init = [init[i] + move[i] for i in range(dimensions)]
            moves += 1

        steps.append(moves)

    mean_steps = np.mean(steps)  
    max_steps = np.max(steps)  
    return mean_steps, max_steps
```

```python
def save_results_to_csv(filename: str, results: list[tuple]) -> None:
    '''
    Append random walk results to a CSV file without deleting previous data.

    Parameters:
    -----------
    filename : str
        Name of the CSV file
    results : list of tuples
        List of (dimension, size, mean_steps, max_steps)
    
    Returns:
    --------
    None
    '''
    file_exists = False
    try:
        with open(filename, 'r'):
            file_exists = True 
    except FileNotFoundError:
        pass  

    with open(filename, mode='a', newline='') as file:
        writer = csv.writer(file)
        if not file_exists:  
            writer.writerow(["Dimension", "Size", "Mean Steps", "Max Steps"])
        writer.writerows(results)
```

```python
def execution(dimensions: int, file_name: str) -> None:
    '''
    Function that executes the random walk in N dimensions

    Parameters:
    -----------
    dimensions : int
        Number of dimensions
    file_name : str
        Name of the file to save the results

    Returns:
    --------
    None
    '''
    results = []
    size = [2**x for x in range(2, 11)]
    print(f"\nTesting {dimensions}D random walk...\n" + "="*40)

    for s in size:
        # Definir los límites para la simulación en N dimensiones
        a_position = tuple(0 for _ in range(dimensions))
        b_position = tuple(s for _ in range(dimensions))

        # Ejecutar la simulación
        mean_steps, max_steps = random_walk_ND(1_000, a_position, b_position)

        # Guardar los resultados
        results.append((dimensions, s, mean_steps, max_steps))

        # Mostrar resultados parciales
        print(f'''\nProblem size: {s} (Dimension: {dimensions}D)
        -----------------------------------
        Mean steps: {mean_steps}
        Max steps: {max_steps}\n''') 

    # Guardar los resultados en un archivo CSV
    save_results_to_csv(file_name, results)
```
